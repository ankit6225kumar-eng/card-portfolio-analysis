```python
g = df.groupby('Attrition_Flag').agg(customers=('CLIENTNUM', 'count'),
                                      spend=('Total_Trans_Amt', 'sum'),
                                      txns=('Total_Trans_Ct', 'sum'))
g['avg_txns'] = g['txns'] / g['customers']
g['avg_ticket'] = g['spend'] / g['txns']
g['avg_spend'] = g['spend'] / g['customers']
g
```

```python
E = g.loc['Existing Customer']
A = g.loc['Attrited Customer']

gap = E['avg_spend'] - A['avg_spend']
freq_effect = (E['avg_txns'] - A['avg_txns']) * (E['avg_ticket'] + A['avg_ticket']) / 2
ticket_effect = (E['avg_ticket'] - A['avg_ticket']) * (E['avg_txns'] + A['avg_txns']) / 2

print('Spend gap per customer:', round(gap))
print('Due to fewer transactions:', round(freq_effect), f'({freq_effect / gap:.0%} of the gap)')
print('Due to ticket size:', round(ticket_effect), f'({ticket_effect / gap:.0%} of the gap)')
print('Check, the two add up to:', round(freq_effect + ticket_effect))
```

```python
df.groupby('Attrition_Flag')[['Total_Ct_Chng_Q4_Q1', 'Months_Inactive_12_mon', 'Total_Trans_Ct']].mean().round(2)
```

```python
rule = (df['Total_Ct_Chng_Q4_Q1'] < 0.7) & (df['Months_Inactive_12_mon'] >= 3)

overall = df['Attrited'].mean()
flagged = df.loc[rule, 'Attrited'].mean()

print('Customers flagged:', rule.sum())
print(f'Overall attrition: {overall:.1%}')
print(f'Attrition among flagged customers: {flagged:.1%}')
print(f'Lift: {flagged / overall:.1f}x')
```

```python
rows = []
for ratio in [0.5, 0.6, 0.7, 0.8]:
    for months in [2, 3, 4]:
        r = (df['Total_Ct_Chng_Q4_Q1'] < ratio) & (df['Months_Inactive_12_mon'] >= months)
        rows.append({
            'ratio_below': ratio,
            'months_inactive_min': months,
            'customers': r.sum(),
            'attrition_pct': round(df.loc[r, 'Attrited'].mean() * 100, 1),
            'lift': round(df.loc[r, 'Attrited'].mean() / overall, 1),
            'leavers_caught_pct': round(df.loc[r, 'Attrited'].sum() / df['Attrited'].sum() * 100, 1)
        })
pd.DataFrame(rows)
```

```python
import numpy as np

at_risk = (df['Total_Ct_Chng_Q4_Q1'] < 0.6) & (df['Months_Inactive_12_mon'] >= 2)
existing = df['Attrition_Flag'] == 'Existing Customer'
high_value = df['Total_Trans_Amt'] >= df['Total_Trans_Amt'].median()

conditions = [
    ~existing,
    existing & at_risk & high_value,
    existing & at_risk & ~high_value,
    existing & ~at_risk & high_value,
]
names = ['Already Left', 'At-Risk High Value', 'At-Risk Low Value', 'Active High Value']
df['Segment'] = np.select(conditions, names, default='Active Low Value')
```

```python
seg = df.groupby('Segment').agg(
    customers=('CLIENTNUM', 'count'),
    avg_spend=('Total_Trans_Amt', 'mean'),
    avg_txns=('Total_Trans_Ct', 'mean'),
    total_spend=('Total_Trans_Amt', 'sum')
)
seg['share_of_customers_pct'] = seg['customers'] / len(df) * 100
seg.round(1)
```

```python
active_spend = df.loc[existing, 'Total_Trans_Amt'].sum()
at_risk_spend = df.loc[existing & at_risk, 'Total_Trans_Amt'].sum()

print('Existing customers flagged at risk:', (existing & at_risk).sum())
print(f'Their share of existing-customer spend: {at_risk_spend / active_spend:.1%}')
```
